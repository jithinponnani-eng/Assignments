```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic=sns.load_dataset("titanic")
```

```python
# Countplot: Passenger Count by Embarked Port
```

```python

plt.figure(figsize=(8,6))
sns.countplot(data=titanic, x='embarked', hue='sex', palette='Set2')

plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- Southampton (S) has the highest number of passengers.")
print("- Male passengers outnumber female passengers at all embarkation ports")
```

```python
# Swarmplot: Age Distribution across Classes
```

```python
plt.figure(figsize=(10,6))
sns.swarmplot(
    data=titanic,
    x='pclass',
    y='age',
    hue='sex',
    palette='Set1',
    size=3
)

plt.title("Age Distribution across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- First-class passengers generally include more older individuals.")
print("- Third class contains a wider spread of younger passengers.\n")
```

```python
# 3. Pie Chart: Gender Distribution
```

```python
gender_counts = titanic['sex'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    explode=(0.05, 0),
    startangle=90,
    colors=['skyblue', 'lightpink']
)

plt.title("Gender Distribution of Passengers")
plt.show()

print("Observation:")
print("- Male passengers form the majority of the dataset.")
print("- The exploded slice highlights the larger gender group.\n")
```

```python
 # 5. Violin Plot: Fare by Class and Gender
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(
    data=titanic,
    x='pclass',
    y='fare',
    hue='sex',
    split=True,
    palette='Pastel1'
)

plt.title("Fare Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- First-class passengers paid significantly higher fares.")
print("- Fare distributions vary more in first class than in lower classes.\n")
```

```python
# 6. Scatter Plot: Age vs Fare
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=titanic,
    x='age',
    y='fare',
    hue='survived',
    palette='viridis'
)

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()

print("Observation:")
print("- Higher fare passengers were more likely to survive.")
print("- Most passengers paid lower fares regardless of age.\n")
```
